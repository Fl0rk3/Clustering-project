```python
import pandas as pd
```

```python
league_data = pd.read_csv('leagues.csv', index_col=0)
```

```python
league_data
```

```python
league_data['Foreign_Player_Shaer'] = league_data['Foreigners'] / league_data['Players_no']
league_data['Players_per_club'] = league_data['Players_no'] / league_data['Club_no']
league_data['Value_per_club'] = league_data['Total_value'] / league_data['Club_no']
```

```python
league_data
```

```python
new_order = [
    "Country",
    "League_name",
    "League_tier",
    "Goals_per_match",
    "Foreign_Player_Shaer",
    "Players_per_club",
    "Value_per_club"
]

fixed_df = league_data[new_order]

fixed_df
```

```python
fixed_df.to_csv('leagues_normalized.csv', index=True)
```
